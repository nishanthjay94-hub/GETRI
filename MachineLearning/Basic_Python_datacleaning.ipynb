```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#to ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\Nishanth\Downloads\employee_data.csv")
```

```python
df


```

```python
#Cleaning 1 : Fill Missing Age using Mean
df["Age"]
df["Age"] = df["Age"].fillna(df["Age"].mean())

df
```

```python
#Cleaning 2 : Fill Salary using Median
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df.head()
```

```python
#Cleaning 3 : Fill Department using Mode
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)
df.head()
```

```python
#Cleaning 4 : Fill Performance Score
df["PerformanceScore"] = df["PerformanceScore"].fillna(
    df["PerformanceScore"].mean()
)
df.head()
```

```python
#Cleaning 5 : Convert Date Column
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])

df.head()
```

```python
#extract year month and date 
df["JoiningYear"] = df["JoiningDate"].dt.year
df["JoiningMonth"] = df["JoiningDate"].dt.month
df["JoiningDay"] = df["JoiningDate"].dt.day

df.head()
```

```python
#Cleaning  : Split Contact Column
df[["Email","Phone"]] = df["Contact"].str.split("|",expand=True)

df.head()
```

```python
#Drop Contact Column
df.drop("Contact",axis=1,inplace=True)
df.head()
```

```python
#Cleaning 9 : Create New Salary Bonus Column
#Bonus = 10%
df["Bonus"] = df["Salary"] * 0.10
df.head()
```

```python
#calculating salary after Bonus
df["TotalSalary"] = df["Salary"] + df["Bonus"]
df.head()
```

```python
#Cleaning 11 : Experience Category

def experience(exp):

    if exp <= 2:
        return "Junior"

    elif exp <=5:
        return "Mid"

    else:
        return "Senior"

df["Level"] = df["Experience"].apply(experience)
```

```python
df.head()
```

```python
#Upper Case Department
df["Department"] = df["Department"].str.upper()
df.head()
```

```python
#Lower Case Email
df["Email"] = df["Email"].str.lower()
df.head()
```

```python
#Employee Initial
df["Initial"] = df["Name"].str[0]
df.head()
```

```python
#First Name and last name
df["FirstName"] = df["Name"].str.split().str[0]
df["LastName"] = df["Name"].str.split().str[1]

df.head(5)
```

```python
#Renaming preformance score to rating
df.rename(columns={"PerformanceScore":"Rating"},inplace=True)
df.head(5)
```

```python
#Sort by Salary
df.sort_values("Salary",ascending=False)
```

```python
#Filter IT Employees
df[df["Department"]=="IT"]
```

```python
#Employees Salary > 50000
df[df["Salary"]>50000]
```

```python
#Duplicate Check
df.duplicated()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df.head()
```

```python
#Check Unique Departments
df["Department"].unique()
```

```python
#save the cleaned dataset
df.to_csv("employee_cleaned.csv",index=False)
```
