```python
#pie

import matplotlib.pyplot as plt
import pandas as pd
data = pd.read_csv(r"C:\Users\Nishanth\Documents\Data_Visualization\Resto_Data.csv")
total_bills_by_day=data.groupby('day')['total_bill'].sum()
total_bills_by_day
explode_data = [0.1 if day== 'Sun' else 0 for day in total_bills_by_day.index]
plt.pie(total_bills_by_day,labels=total_bills_by_day.index,
        autopct='%1.1f%%',explode=explode_data)
plt.title('total bills by day')
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
data = pd.read_csv(r"C:\Users\Nishanth\Documents\Data_Visualization\Resto_Data.csv")

# Create box plot
plt.boxplot(data['total_bill'], vert=True, patch_artist=True, 
            boxprops=dict(facecolor="lightblue"))

# Customize plot
plt.title("Box Plot of Total Bill Amounts")
plt.ylabel("Total Bill ($)")
plt.grid(axis="y", linestyle="--", alpha=0.7)  # Adds grid lines for better visibility

# Show plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
# Load dataset
data = pd.read_csv(r"C:\Users\Nishanth\Documents\Data_Visualization\Resto_Data.csv")
# Compute Q1 and Q3
Q1 = data['total_bill'].quantile(0.25)
Q3 = data['total_bill'].quantile(0.75)
# Calculate Interquartile Range (IQR)
IQR = Q3 - Q1
# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Filter the data (remove outliers)
clean_data = data[(data['total_bill'] >= lower_bound) & (data['total_bill'] <= upper_bound)]
print("Original data size:", len(data))
print("After removing outliers:", len(clean_data))
# Boxplot after removing outliers
plt.boxplot(clean_data['total_bill'], vert=True, patch_artist=True,boxprops=dict(facecolor='lightgreen', alpha=0.7))
plt.title("Box Plot of Total Bill (After Removing Outliers)")
plt.ylabel("Total Bill ($)")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
