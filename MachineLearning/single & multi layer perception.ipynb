```python
# single layer perception

import numpy as np
import tensorflow as tf

# Sample dataset: [Size (sqft), Bedrooms], Price (in $1000s)
X = np.array([[800, 2], [1000, 3], [1200, 3], [1500, 4], [1800, 4]], dtype=np.float32)  # Features
y = np.array([[200], [250], [300], [400], [450]], dtype=np.float32)  # Target (Price)

# Define a single-layer perceptron
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_dim=2)  # Single layer with 1 output neuron
])

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])

# Train the model
model.fit(X, y, epochs=200, verbose=0)

# Evaluate the model
loss, mae = model.evaluate(X, y, verbose=0)
print(f"Mean Absolute Error: {mae:.2f}")

# Predict
predictions = model.predict(X)
print("Predictions (in $1000s):")
print(predictions)
```
